```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset

df = pd.read_csv('UNSW F12024.csv')
```

# Albert Park Grand Prix Circuit

- There are 4 unique tracks in the dataset
- Track 0 corresponds to the Albert Park Grand Prix Circuit

```python
# Count how many data for each different track
df['M_TRACKID'].value_counts()
```

```python
# Examples of each track

track0_example = df[df["SESSION_GUID"] == '1BF21BECE1C29DBFE0631718000AACA5']
track32_example = df[df["SESSION_GUID"] == '229ABDD3A294903EE0631618000A8D9F']
track29_example = df[df["SESSION_GUID"] == '2908CC0E4F9C27E3E0631718000A9F4A']
track15_example = df[df["SESSION_GUID"] == '265E99649E788412E0631218000A43C1']

# Plotting the 4 unique tracks

fig, axes = plt.subplots(1, 4, figsize=(20, 5))  # 1 row, 4 columns

axes[0].scatter(track0_example['M_WORLDPOSITIONX_1'], track0_example['M_WORLDPOSITIONY_1'], color="red")
axes[0].set_title("Track 0")

axes[1].scatter(track32_example['M_WORLDPOSITIONX_1'], track32_example['M_WORLDPOSITIONY_1'], color="blue")
axes[1].set_title("Track 32")

axes[2].scatter(track29_example['M_WORLDPOSITIONX_1'], track29_example['M_WORLDPOSITIONY_1'], color="green")
axes[2].set_title("Track 29")

axes[3].scatter(track15_example['M_WORLDPOSITIONX_1'], track15_example['M_WORLDPOSITIONY_1'], color="yellow")
axes[3].set_title("Track 15")

plt.tight_layout()
plt.show()
```
